```python
import re
```

```python
string= 'Artificial Intelligence 999 and Machine Learning' 
pattern=r"\d{3}"
result=re.search(pattern,string=string)
print(string[result.span()[0]:result.span()[1]])
```

```python
pattern=re.compile(r'end')
my_string="""write Anything 123 %4566sds endhere
end2
""" 
matches=pattern.finditer(my_string)
for match in matches:
    print(match)
```

```python
pattern=re.compile(r'%\d{4}\s')
my_string="""write Anything 123 %4566 sds endhere
end2
""" 
matches=pattern.finditer(my_string)
for match in matches:
    print(match)
```

```python
emailP=pattern=re.compile(r'\[[alpha]]@.\.')
matches=emailP.finditer("raman@a.com")
for match in matches:
    print(match)
```
